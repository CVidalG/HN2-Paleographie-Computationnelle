```python
from ultralytics import FastSAM

# Define an inference source
source = "/Users/nostradamus/Documents/Chartes/Cours_ENC/Github/HN2-Paleographie-Computationnelle/lesson2-enluminures/Data/DataBnFSegmentation2019/MiradorAnnotations/images/Giuseppe_Castiglione_05.jpg"

# Create a FastSAM model
model = FastSAM("FastSAM-s.pt")  # or FastSAM-x.pt
```

```python
# Run inference on an image
everything_results = model(source, device="cpu", retina_masks=True, imgsz=1024, conf=0.4, iou=0.9, save=True)
```
